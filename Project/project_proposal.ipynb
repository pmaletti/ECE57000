## Slide 1: Track Selection & Problem Formulation

**Track Selection:** Novel Research

**Problem Formulation:**
Diffusion models suffer from massive inference bottlenecks. Current structured pruning solutions are inefficient, requiring either costly end-to-end retraining or memory-intensive full Hessian inversions to recover generative quality.

**Proposed Solution:**
I will develop **Isometric Spectral Pruning**, a lightweight framework that uses local spectral isometry as a geometric saliency metric to robustly select and prune redundant structures (e.g., attention heads, MLPs) without Hessian inversion. I will systematically reproduce the SDXL structured pruning results of OBS-Diff and EcoDiff and benchmark them against this novel approach.

**Proposed Setup:** 
* I will prune the SDXL U-Net architecture, targeting up to a 40% parameter reduction using a local isometry-based correction.
* This introduces a novel geometric correction for structured pruning that uniquely preserves the underlying structure of the model's activation and weight spaces.
* By utilizing pre-trained SDXL baselines, the primary computational effort is isolated to the zero-shot proxy evaluation, making it highly executable within the semester.

## Slide 2: Literature Review - Paper 1

**OBS-Diff: Accurate Pruning For Diffusion Models in One-Shot [1]**
*(ICLR 2026)*

*   **Main Contribution:**
    OBS-Diff adapts the classic Optimal Brain Surgeon (OBS) [4] framework to text-to-image diffusion models. It introduces a training-free, one-shot pruning method that utilizes a timestep-aware Hessian construction. By weighting earlier denoising steps more heavily, it prevents error accumulation across the iterative diffusion process. This briefly summarizes the main contribution of the paper.
*   **Relevance to Proposed Project:**
    This paper serves as the primary "training-free" baseline. I will reproduce its structured pruning pipeline on SDXL. OBS-Diff relies on computationally expensive full Hessian inversions, it highlights the computational bottleneck that my proposed geometric framework aims to resolve. This briefly explains how the specific paper relates to the project.

## Slide 3: Literature Review - Paper 2

**Learnable Sparsity for Vision Generative Models (EcoDiff) [2]**
*(ICLR 2026)*

*   **Main Contribution:**
    EcoDiff presents a model-agnostic, end-to-end structural pruning framework that **learns** a differentiable neuron mask. To manage the massive memory requirements of backpropagating across the entire denoising trajectory, it introduces time-step gradient checkpointing, enabling the efficient structural pruning of models like SDXL. This briefly summarizes the main contribution of the paper.
*   **Relevance to Proposed Project:**
    EcoDiff acts as the "learnable sparsity" baseline. Reproducing its SDXL pruned model provides a direct comparative benchmark. It will allow me to evaluate whether my isometric pruning proxy can match the structural integrity preserved by complex, gradient-based trajectory masking. This briefly explains how the specific paper relates to the project.

## Slide 4: Literature Review - Paper 3

**Diffusion Pruning for Rapidly and Robustly Selecting Global Correspondences Using Local Isometry [3]**
*(ACM Transactions on Graphics, 2014)*

*   **Main Contribution:**
    This paper introduces a novel, rapid network pruning methodology that prioritizes the preservation of local geometric structures. By evaluating local isometry, it robustly selects global correspondences to identify redundant parameters without iterative training loops or heavy second-order derivative calculations. It is crucial to note that "diffusion" in this context refers to **diffusion maps** (a manifold learning technique), *not* generative diffusion models. This graphics paper focuses on estimating dense correspondences between non-rigid 3D shapes (i.e., mapping the same object in two different poses). The authors use diffusion maps to robustly evaluate local isometry, allowing them to rapidly prune incorrect global correspondence candidates between the shapes. This briefly summarizes the main contribution of the paper.
*   **Relevance to Proposed Project:**
    This work forms the theoretical foundation of my proposed geometric proxy. While originally designed to evaluate 3D meshes, I will adapt this diffusion map and local isometry framework to evaluate the internal data manifold of the SDXL U-Net. Instead of pruning incorrect physical poses, I will use this mathematical approach to rapidly identify and prune redundant cross-attention and MLP structures, creating a zero-shot proxy that avoids the heavy calculations of OBS-Diff or EcoDiff.

## Slide 5: Evaluation Metric

**Proposed Metrics:** Performance will be objectively quantified by evaluating the 30% structurally pruned models across five standard absolute metrics: FID [5], CLIP Score [6], Inception Score (IS) [7], Image Reward [8], and SSIM [9].

* **Propose & Justify:**
    While SSIM (measured against base model generations using identical seeds) and IS measure basic structural similarity and diversity, absolute FID (feature distribution distance) and CLIP Score (text-image alignment) are quantifiable and accurately evaluate if the generative manifold is preserved. Incorporating Image Reward ensures human-preference alignment. Excelling across these computable metrics at 30% sparsity proves the problem is effectively solved by demonstrating that the isometric framework identified redundant structures as effectively as the heavier baselines.

* **Contextualize (The 4 Tiers):** Assuming evaluation on a standard benchmark (e.g., MS-COCO zero-shot), performance is defined by the absolute scores below:

| Tier | FID | CLIP Score | IS | Image Reward | SSIM | Qualitative Result |
| :--- | :--- | :--- | :--- | :--- | :--- | :--- |
| **Excellent** | ≤ 33 | ≥ 0.31 | ≥ 40 | ≥ 1.0 | ≥ 0.90 | Indistinguishable from baseline; out-competes OBS-Diff & EcoDiff. |
| **Good** | 34 - 39 | 0.28 - 0.309 | 35 - 39.9 | 0.5 - 0.99 | 0.75 - 0.89 | Minor statistical degradation; high fidelity and alignment maintained. |
| **Okay** | 40 - 45 | 0.25 - 0.279 | 25 - 34.9 | 0.0 - 0.49 | 0.60 - 0.74 | Noticeable artifacting, reduced diversity, and drop in prompt adherence. |
| **Poor** | > 45 | < 25 | < 0.25 | < 0.0 | < 0.60 | Severe mode collapse; catastrophic structural damage compared to baselines. |
